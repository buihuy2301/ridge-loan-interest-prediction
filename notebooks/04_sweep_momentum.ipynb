Ba công thức momentum và khởi động lại thích nghi. Chương 5 của báo cáo.

```python
import sys; sys.path.insert(0, "..")
import numpy as np, pandas as pd
pd.set_option("display.width", 160)
```

```python
from src.dataset import load_processed, SWEEP
from src.experiment import BUILDERS, ExperimentGroup, run_group, summary_table
from src.figures import convergence_pair
obj, *_ = load_processed("../data/processed", SWEEP)

groups = {}
for name in ("momentum-formula", "momentum-restart"):
    title, build = BUILDERS[name]
    groups[name] = run_group(ExperimentGroup(name, title, build), obj, "../results/raw")
    convergence_pair(groups[name], name, title=title, out_dir="../results/figures")
```

```python
for name, recs in groups.items():
    df = pd.DataFrame(summary_table(recs))
    df["cấu hình"] = [r.label for r in recs]
    print(f"=== {name} ===")
    print(df[["cấu hình", "status", "iters_to_1e-06", "seconds_to_1e-06", "final_gap"]].to_string(index=False))
    print()
```

Momentum tối ưu theo lý thuyết so với hằng số 0,9 mà đề bài gợi ý.

```python
beta_opt = (np.sqrt(obj.kappa) - 1) / (np.sqrt(obj.kappa) + 1)
print(f"kappa = {obj.kappa:.1f}  ->  beta* = {beta_opt:.4f}, so voi 0.9 duoc goi y")
```
